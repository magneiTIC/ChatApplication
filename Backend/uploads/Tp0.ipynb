```python
import numpy as np
import pandas as pd
from tensorflow.keras import datasets
from tensorflow.keras.preprocessing import sequence
```

```python
imdb = datasets.imdb
(X_train, y_train),(X_test, y_test) = imdb.load_data(num_words=10000)
X_train.shape, y_train.shape
```

```python
X_train = sequence.pad_sequences(X_train, value = 0, padding = 'post', maxlen = 256)
X_test = sequence.pad_sequences(X_test, value = 0, padding = 'post', maxlen = 256)
```

```python
print(len(X_train[0]), len(X_train[5]))
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, SimpleRNN
```

```python
model = Sequential()
#Transformation(Embedding)
model.add(Embedding(input_dim = 10000, input_length = 256, output_dim = 32))
#MLP(SimpleRNN)
model.add(SimpleRNN(units = 8, return_sequences = True))
model.add(SimpleRNN(units = 8, return_sequences = False))
#Sortie
model.add(Dense(units = 1, activation = 'sigmoid'))
```

```python
model.compile(optimizer = 'rmsprop', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
model.fit(x = X_train, y = y_train, validation_data = (X_test, y_test), epochs = 10, batch_size = 128)
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy : ', model.evaluate(X_test, y_test)[1])
```
